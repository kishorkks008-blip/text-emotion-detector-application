```python
from textblob import TextBlob
from collections import Counter
import re
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
text = input("Enter your text: ")

print("\nYour Text:")
print(text)
```

```python
# Clean text
clean_text = re.sub(r'[^a-zA-Z\s]', '', text.lower())

# Words
words = clean_text.split()

# Basic statistics
word_count = len(words)
character_count = len(text)

# TextBlob sentiment
blob = TextBlob(text)
polarity = blob.sentiment.polarity
subjectivity = blob.sentiment.subjectivity

print("\n===== TEXT ANALYSIS =====")
print("Number of words      :", word_count)
print("Number of characters :", character_count)
print("Polarity             :", round(polarity, 2))
print("Subjectivity         :", round(subjectivity, 2))
```

```python
def detect_emotion(text):
    text = text.lower()

    emotion_words = {
        "Joy": [
            "happy", "joy", "excited", "wonderful", "great",
            "love", "excellent", "amazing", "fun", "success"
        ],
        "Sadness": [
            "sad", "cry", "crying", "lonely", "depressed",
            "unhappy", "hurt", "pain", "lost", "upset"
        ],
        "Anger": [
            "angry", "hate", "furious", "annoyed", "mad",
            "irritated", "rage", "frustrated"
        ],
        "Fear": [
            "afraid", "fear", "scared", "worried", "danger",
            "nervous", "terrified", "panic"
        ],
        "Surprise": [
            "wow", "surprise", "surprised", "unexpected",
            "shocked", "amazing", "suddenly"
        ]
    }

    scores = {}

    for emotion, keywords in emotion_words.items():
        score = 0

        for word in keywords:
            score += text.count(word)

        scores[emotion] = score

    detected_emotion = max(scores, key=scores.get)

    if scores[detected_emotion] == 0:
        detected_emotion = "Neutral"

    return detected_emotion, scores


emotion, emotion_scores = detect_emotion(text)

print("\n===== EMOTION RESULT =====")
print("Detected Emotion:", emotion)
```

```python
labels = list(emotion_scores.keys())
values = list(emotion_scores.values())

plt.figure(figsize=(8, 5))
plt.bar(labels, values)

plt.title("Emotion Analysis")
plt.xlabel("Emotion")
plt.ylabel("Score")

plt.show()
```

```python
if clean_text.strip():

    wordcloud = WordCloud(
        width=900,
        height=500,
        background_color="white"
    ).generate(clean_text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud")

    plt.show()
```

```python
print("\n" + "=" * 40)
print("       TEXT EMOTION REPORT")
print("=" * 40)

print("Text              :", text)
print("Words             :", word_count)
print("Characters        :", character_count)
print("Polarity          :", round(polarity, 2))
print("Subjectivity      :", round(subjectivity, 2))
print("Detected Emotion  :", emotion)

if polarity > 0:
    sentiment = "Positive"
elif polarity < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print("Sentiment         :", sentiment)

print("=" * 40)
```
